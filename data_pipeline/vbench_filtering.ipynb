```python
import json
import os
```

```python
base = 'evaluation_results_edited/' # edited video
base_original = 'evaluation_results_original/' # original video
domain = ['aesthetic','motion', 'quality', 'subject', 'dynamic', 'background'] # 
domain_original = ['aesthetic', 'quality']
```

```python
eval_results = {}
for dom in domain:
    eval_results[dom] = {}
    for i in range(10):
        f = os.listdir(os.path.join(base, dom + '_{}'.format(str(i+1))))
        for f_ in f:
            if 'eval' in f_:
                d = json.load(open(os.path.join(base, dom + '_{}'.format(str(i+1)), f_)))
                d = d[list(d.keys())[0]][1]
                for d_ in d:
                    eval_results[dom][d_['video_path'].split('/')[-1].split('.')[0]] = d_['video_results']
```

```python
eval_results_original = {}
for dom in domain_original:
    eval_results_original[dom] = {}
    for i in range(10):
        f = os.listdir(os.path.join(base_original, dom + '_{}'.format(str(i+1))))
        for f_ in f:
            if 'eval' in f_:
                d = json.load(open(os.path.join(base_original, dom + '_{}'.format(str(i+1)), f_)))
                d = d[list(d.keys())[0]][1]
                for d_ in d:
                    eval_results_original[dom][d_['video_path'].split('/')[-1].split('.')[0]] = d_['video_results']
```
